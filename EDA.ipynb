```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
sns.set_style("whitegrid")
```

```python
df = pd.read_csv("economics_data.csv")
df.shape
```

```python
df["Date"] = pd.to_datetime(df["Date"], format="%d-%m-%Y")
```

```python
# Check missing values in each column

df.isnull().sum()
```

```python
# Check the number of duplicate rows

df.duplicated().sum()
```

```python
df.describe()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Inflation Rate (%)"], bins=20)
plt.title("Distribution of inflation rate")
plt.xlabel("Inflation Rate (%)")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["GDP Growth (%)"], kde=True)
plt.title("Distribution of GDP Growth")
plt.xlabel("GDP Growth (%)")
plt.show()
```

```python
correlation = df.corr(numeric_only=True)
correlation
```

```python
plt.figure(figsize=(9,8))
sns.heatmap(correlation,annot=True,cmap="cool",fmt=".2f",linewidths=0.5)
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x="Interest Rate (%)",
    y="Consumer Spending (Billion USD)",
    data=df
)

plt.title("Interest Rate vs Consumer Spending")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(x="Unemployment Rate (%)", y="GDP Growth (%)",data=df)
plt.title("Unemployment Rate vs GDP Growth")
plt.show()
```

```python
from sklearn.cluster import KMeans

X = df.select_dtypes(include="number")

kmeans = KMeans(n_clusters=3, random_state=42)

df["Cluster"] = kmeans.fit_predict(X)

df.head()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x="GDP Growth (%)",
    y="Inflation Rate (%)",
    hue="Cluster",
    data=df,
    palette="Set1"
)

plt.title("Cluster Analysis")
plt.show()
```

```python
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

df["Year"] = df["Date"].dt.year
```

```python
gdp_year = df.groupby("Year")["GDP Growth (%)"].mean()
```

```python
plt.figure(figsize=(10,5))
plt.plot(gdp_year.index, gdp_year.values, marker='o', linewidth=2)
plt.title("GDP Growth by Year")
plt.xlabel("Year")
plt.ylabel("GDP Growth (%)")
plt.xticks(gdp_year.index)
plt.grid(True)
plt.show()
```

```python
inflation_year = df.groupby("Year")["Inflation Rate (%)"].mean()
plt.figure(figsize=(10,5))
plt.plot(inflation_year.index, inflation_year.values, marker='o', linewidth=2)
plt.title("Inflation Rate by Year")
plt.xlabel("Year")
plt.ylabel("Inflation Rate (%)")
plt.xticks(inflation_year.index)
plt.grid(True)
plt.show()
```

```python
unemployment_year = df.groupby("Year")["Unemployment Rate (%)"].mean()
plt.figure(figsize=(10,5))
plt.plot(unemployment_year.index, unemployment_year.values, marker='o', linewidth=2)
plt.title("Average Unemployment Rate by Year")
plt.xlabel("Year")
plt.ylabel("Unemployment Rate (%)")
plt.xticks(unemployment_year.index)
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

yearly = pd.DataFrame({
    "GDP Growth (%)": gdp_year,
    "Inflation Rate (%)": inflation_year
})

yearly.plot(
    kind="bar",
    figsize=(10,6),
    width=0.8
)

plt.title("Average GDP Growth and Inflation Rate by Year")
plt.xlabel("Year")
plt.ylabel("Percentage (%)")
plt.xticks(rotation=45)
plt.legend(title="Indicators")
plt.tight_layout()
plt.show()
```

```python
# Convert Date column
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# Extract Year
df["Year"] = df["Date"].dt.year

# Calculate yearly average GDP Growth
gdp_year = df.groupby("Year")["GDP Growth (%)"].mean()

# Plot
plt.figure(figsize=(10,5))

plt.plot(gdp_year.index,
         gdp_year.values,
         marker='o',
         linewidth=2)

plt.title("Average GDP Growth by Year")
plt.xlabel("Year")
plt.ylabel("GDP Growth (%)")

plt.xticks(gdp_year.index)

plt.grid(True)

plt.show()
```
